```python
import sys
import time
import hashlib
import json
import platform

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import pearsonr
from scipy.special import erfc

SEED = 2026

# Experiment parameters from the paper
N = 256
BURN_IN = 256

U1 = 3.9
U2 = 3.7
R1 = 0.1
R2 = 0.1
T = 0.3

X0 = 0.1
Y0 = 0.3

NUM_FRAMES = 20
HEIGHT = 64
WIDTH = 64
CHANNELS = 3
DISCRETE_STEPS = 32

np.random.seed(SEED)

print("Python:", sys.version)
print("Platform:", platform.platform())
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Frames:", NUM_FRAMES)
print("Frame dimensions:", (HEIGHT, WIDTH, CHANNELS))
print("Modulus:", N)
```

```python
def generate_chaotic_sequences(
    length=10000,
    burn_in=BURN_IN,
    x0=X0,
    y0=Y0,
    u1=U1,
    u2=U2,
    r1=R1,
    r2=R2
):
    total = length + burn_in

    x = np.zeros(total, dtype=np.float64)
    y = np.zeros(total, dtype=np.float64)

    x[0] = x0
    y[0] = y0

    for i in range(total - 1):
        xn = x[i]
        yn = y[i]

        # Assumed coupled generalized logistic recurrence
        x_next = u1 * xn * (1.0 - xn) + r1 * yn
        y_next = u2 * (yn ** 2) * ((1.0 - yn) ** 2) + r2 * xn

        # Modulo-one folding
        x[i + 1] = x_next % 1.0
        y[i + 1] = y_next % 1.0

    return x[burn_in:], y[burn_in:]


X, Y = generate_chaotic_sequences(length=10000)

print("X length:", len(X))
print("Y length:", len(Y))
print("First 10 X values:", X[:10])
print("First 10 Y values:", Y[:10])
print("Pearson correlation:", np.corrcoef(X, Y)[0, 1])

assert len(X) == 10000
assert len(Y) == 10000
assert np.all((X >= 0) & (X < 1))
assert np.all((Y >= 0) & (Y < 1))
```

```python
def piecewise_cosine_map(values, t=T):
    """
    Four-interval piecewise cosine transformation.

    Assumption:
    Normalize each cosine output to [0, 1] using
    (1 + cos(z)) / 2.
    """
    values = np.asarray(values, dtype=np.float64) % 1.0
    result = np.empty_like(values)

    m1 = (values >= 0.0) & (values < t)
    m2 = (values >= t) & (values < 0.5)
    m3 = (values >= 0.5) & (values < 1.0 - t)
    m4 = (values >= 1.0 - t) & (values < 1.0)

    result[m1] = np.cos(values[m1] / t)
    result[m2] = np.cos(
        (values[m2] - t) / (0.5 - t)
    )
    result[m3] = np.cos(
        (1.0 - values[m3] - (1.0 - 2.0 * t))
        / (0.5 - t)
    )
    result[m4] = np.cos((1.0 - values[m4]) / t)

    result = (1.0 + result) / 2.0
    return np.clip(result, 0.0, 1.0)


CX = piecewise_cosine_map(X)
CY = piecewise_cosine_map(Y)

print("CX range:", CX.min(), CX.max())
print("CY range:", CY.min(), CY.max())
print("Cosine-mapped sequence correlation:",
      np.corrcoef(CX, CY)[0, 1])

plt.figure(figsize=(10, 3))
plt.plot(CX[:500], label="CX")
plt.plot(CY[:500], label="CY", alpha=0.8)
plt.title("First 500 Piecewise-Cosine Outputs")
plt.xlabel("Iteration")
plt.ylabel("Value")
plt.legend()
plt.show()
```

```python
def generate_discrete_keystreams(
    cx,
    cy,
    steps=DISCRETE_STEPS,
    modulus=N
):
    # Use 64 x 64 states for direct RGB frame processing.
    size = HEIGHT * WIDTH

    a = np.resize(
        np.floor(cx * modulus).astype(np.int64), size
    ).reshape(HEIGHT, WIDTH)

    b = np.resize(
        np.floor(cy * modulus).astype(np.int64), size
    ).reshape(HEIGHT, WIDTH)

    # Preserve the initial states
    a = a.copy() % modulus
    b = b.copy() % modulus

    for m in range(steps):
        # Coefficient pattern based on the document.
        if m % 2 == 0:
            am = bm = 1
            cm = tm = 1
            rm = sm = 0
        else:
            am = bm = 0
            cm = tm = 1
            rm = sm = 1

        # Periodic boundary conditions using array rolls.
        a_right = np.roll(a, -1, axis=1)
        b_right = np.roll(b, -1, axis=1)

        new_a = (
            am * b + bm * a + cm * a_right
        ) % modulus

        new_b = (
            rm * a + sm * b + tm * b_right
        ) % modulus

        a, b = new_a.astype(np.int64), new_b.astype(np.int64)

    k1 = a.astype(np.uint8).flatten()
    k2 = b.astype(np.uint8).flatten()

    return k1, k2


K1, K2 = generate_discrete_keystreams(CX, CY)

print("K1 length:", len(K1))
print("K2 length:", len(K2))
print("K1 first 20 bytes:", K1[:20])
print("K2 first 20 bytes:", K2[:20])
print("Keystream cross-correlation:",
      np.corrcoef(K1[:1000], K2[:1000])[0, 1])

assert K1.dtype == np.uint8
assert K2.dtype == np.uint8
```

```python
def generate_synthetic_video(
    num_frames=NUM_FRAMES,
    height=HEIGHT,
    width=WIDTH
):
    video = np.zeros(
        (num_frames, height, width, 3),
        dtype=np.uint8
    )

    yy, xx = np.indices((height, width))

    for f in range(num_frames):
        # Explicitly chosen deterministic test pattern.
        red = (
            3 * xx + 2 * yy + 7 * f
            + 25 * np.sin((xx + f) / 7.0)
        ) % 256

        green = (
            2 * xx + 4 * yy + 5 * f
            + 25 * np.sin((yy + 2 * f) / 9.0)
        ) % 256

        blue = (
            4 * xx + 3 * yy + 11 * f
            + 25 * np.sin((xx + yy + f) / 11.0)
        ) % 256

        video[f, :, :, 0] = red.astype(np.uint8)
        video[f, :, :, 1] = green.astype(np.uint8)
        video[f, :, :, 2] = blue.astype(np.uint8)

    return video


original_video = generate_synthetic_video()

print("Video shape:", original_video.shape)
print("Video dtype:", original_video.dtype)
print("Video size in bytes:", original_video.nbytes)

plt.figure(figsize=(5, 5))
plt.imshow(original_video[0])
plt.title("Original Synthetic RGB Frame 0")
plt.axis("off")
plt.show()
```

```python
def frame_masks(k1, k2, frame_index):
    size = HEIGHT * WIDTH * CHANNELS

    offset = frame_index * size

    mask1 = np.resize(
        np.roll(k1, offset % len(k1)), size
    ).reshape(HEIGHT, WIDTH, CHANNELS)

    mask2 = np.resize(
        np.roll(k2, offset % len(k2)), size
    ).reshape(HEIGHT, WIDTH, CHANNELS)

    return mask1, mask2


def encrypt_frame(frame, k1, k2, frame_index):
    frame = np.asarray(frame, dtype=np.uint8)
    encrypted = frame.copy()

    mask1, mask2 = frame_masks(k1, k2, frame_index)

    # Pairwise row operations.
    for row in range(0, HEIGHT - 1, 2):
        odd_plain = frame[row]
        even_plain = frame[row + 1]

        odd_encrypted = odd_plain ^ mask1[row] ^ mask2[row]
        even_encrypted = even_plain ^ odd_encrypted

        encrypted[row] = odd_encrypted
        encrypted[row + 1] = even_encrypted

    # For an unpaired final row, if height is odd.
    if HEIGHT % 2:
        row = HEIGHT - 1
        encrypted[row] = (
            frame[row] ^ mask1[row] ^ mask2[row]
        )

    return encrypted


def decrypt_frame(encrypted_frame, k1, k2, frame_index):
    encrypted_frame = np.asarray(
        encrypted_frame, dtype=np.uint8
    )
    recovered = encrypted_frame.copy()

    mask1, mask2 = frame_masks(k1, k2, frame_index)

    for row in range(0, HEIGHT - 1, 2):
        odd_encrypted = encrypted_frame[row]
        even_encrypted = encrypted_frame[row + 1]

        even_plain = even_encrypted ^ odd_encrypted
        odd_plain = odd_encrypted ^ mask1[row] ^ mask2[row]

        recovered[row] = odd_plain
        recovered[row + 1] = even_plain

    if HEIGHT % 2:
        row = HEIGHT - 1
        recovered[row] = (
            encrypted_frame[row] ^ mask1[row] ^ mask2[row]
        )

    return recovered


# Validate a single frame first.
test_encrypted = encrypt_frame(
    original_video[0], K1, K2, 0
)
test_recovered = decrypt_frame(
    test_encrypted, K1, K2, 0
)

assert np.array_equal(original_video[0], test_recovered)

print("Single-frame encryption/decryption test: PASS")

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(original_video[0])
plt.title("Original")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(test_encrypted)
plt.title("Encrypted")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(test_recovered)
plt.title("Recovered")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
encrypted_video = np.empty_like(original_video)
recovered_video = np.empty_like(original_video)

start = time.perf_counter()

for f in range(NUM_FRAMES):
    encrypted_video[f] = encrypt_frame(
        original_video[f], K1, K2, f
    )

encryption_time = time.perf_counter() - start

start = time.perf_counter()

for f in range(NUM_FRAMES):
    recovered_video[f] = decrypt_frame(
        encrypted_video[f], K1, K2, f
    )

decryption_time = time.perf_counter() - start

exact_recovery = np.array_equal(
    original_video, recovered_video
)

print(f"Encryption time: {encryption_time:.6f} seconds")
print(f"Encryption time/frame: "
      f"{encryption_time / NUM_FRAMES:.6f} seconds")
print(f"Decryption time: {decryption_time:.6f} seconds")
print("Exact recovery:", exact_recovery)

assert exact_recovery, "Decryption did not recover the original video."

print("Full-video round-trip test: PASS")
```

```python
def shannon_entropy(frame):
    values, counts = np.unique(
        np.asarray(frame, dtype=np.uint8),
        return_counts=True
    )
    probabilities = counts / counts.sum()

    return float(
        -np.sum(probabilities * np.log2(probabilities))
    )


entropy_rows = []

for f in range(NUM_FRAMES):
    entropy_rows.append({
        "Frame": f,
        "Original entropy": shannon_entropy(original_video[f]),
        "Encrypted entropy": shannon_entropy(encrypted_video[f]),
        "Recovered entropy": shannon_entropy(recovered_video[f])
    })

entropy_df = pd.DataFrame(entropy_rows)

display(entropy_df.head(5).round(6))

plt.figure(figsize=(9, 4))
plt.plot(
    entropy_df["Frame"],
    entropy_df["Original entropy"],
    marker="o",
    label="Original"
)
plt.plot(
    entropy_df["Frame"],
    entropy_df["Encrypted entropy"],
    marker="s",
    label="Encrypted"
)
plt.xlabel("Frame")
plt.ylabel("Shannon entropy (bits/symbol)")
plt.title("Original vs. Encrypted Frame Entropy")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
def npcr(frame_a, frame_b):
    a = np.asarray(frame_a, dtype=np.uint8)
    b = np.asarray(frame_b, dtype=np.uint8)

    return 100.0 * np.mean(a != b)


def uaci(frame_a, frame_b):
    a = np.asarray(frame_a, dtype=np.int16)
    b = np.asarray(frame_b, dtype=np.int16)

    return 100.0 * np.mean(np.abs(a - b)) / 255.0


npcr_values = []
uaci_values = []

for f in range(NUM_FRAMES - 1):
    npcr_values.append(
        npcr(encrypted_video[f], encrypted_video[f + 1])
    )
    uaci_values.append(
        uaci(encrypted_video[f], encrypted_video[f + 1])
    )

mean_npcr = float(np.mean(npcr_values))
mean_uaci = float(np.mean(uaci_values))

print(f"Mean adjacent-frame NPCR: {mean_npcr:.6f}%")
print(f"Mean adjacent-frame UACI: {mean_uaci:.6f}%")

print("\nFirst five NPCR measurements:")
print(np.round(npcr_values[:5], 6))

print("\nFirst five UACI measurements:")
print(np.round(uaci_values[:5], 6))
```

```python
def frame_correlation(frame_a, frame_b):
    a = np.asarray(frame_a, dtype=np.float64).flatten()
    b = np.asarray(frame_b, dtype=np.float64).flatten()

    return float(np.corrcoef(a, b)[0, 1])


correlation_before = frame_correlation(
    original_video[0], original_video[1]
)

correlation_after = frame_correlation(
    encrypted_video[0], encrypted_video[1]
)

print(f"Correlation before encryption: {correlation_before:.6f}")
print(f"Correlation after encryption:  {correlation_after:.6f}")

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].scatter(
    original_video[0].flatten(),
    original_video[1].flatten(),
    s=2
)
axes[0].set_title(
    f"Original frames (r = {correlation_before:.4f})"
)
axes[0].set_xlabel("Frame 0 byte values")
axes[0].set_ylabel("Frame 1 byte values")

axes[1].scatter(
    encrypted_video[0].flatten(),
    encrypted_video[1].flatten(),
    s=2
)
axes[1].set_title(
    f"Encrypted frames (r = {correlation_after:.4f})"
)
axes[1].set_xlabel("Encrypted frame 0 byte values")
axes[1].set_ylabel("Encrypted frame 1 byte values")

plt.tight_layout()
plt.show()
```

```python
def monobit_test(bit_sequence):
    bits = np.asarray(bit_sequence, dtype=np.uint8).flatten()

    if len(bits) == 0:
        raise ValueError("Bit sequence cannot be empty.")

    # Map 0 -> -1 and 1 -> +1.
    signed_bits = 2 * bits.astype(np.int64) - 1

    n = len(signed_bits)
    statistic = abs(np.sum(signed_bits)) / np.sqrt(n)

    p_value = float(erfc(statistic / np.sqrt(2.0)))

    return p_value


def bytes_to_bits(byte_array):
    return np.unpackbits(
        np.asarray(byte_array, dtype=np.uint8)
    )


# Use the first 1000 bytes from each discrete stream.
combined_keystream = np.concatenate([
    K1[:1000],
    K2[:1000]
])

base_bits = bytes_to_bits(combined_keystream)

monobit_rows = []

for bit_length in [1024, 4096, 16384]:
    # Match the document's repeat-to-length procedure.
    repeats = int(np.ceil(bit_length / len(base_bits)))
    bits = np.tile(base_bits, repeats)[:bit_length]

    p_value = monobit_test(bits)

    monobit_rows.append({
        "Bit length": bit_length,
        "p-value": p_value,
        "Threshold": 0.01,
        "Result": "PASS" if p_value > 0.01 else "FAIL"
    })

monobit_df = pd.DataFrame(monobit_rows)
display(monobit_df.round(6))
```

```python
def serialize_video(video):
    return np.ascontiguousarray(video).tobytes()


def create_payload_digest(encrypted_video):
    return hashlib.sha256(
        serialize_video(encrypted_video)
    ).hexdigest()


def verify_payload_digest(encrypted_video, expected_digest):
    actual_digest = create_payload_digest(encrypted_video)
    return actual_digest == expected_digest


# Demonstration NFT metadata.
nft_metadata = {
    "token_id": 1001,
    "standard": "ERC-721",
    "payload_format": "encrypted RGB frame sequence",
    "frame_count": NUM_FRAMES,
    "frame_width": WIDTH,
    "frame_height": HEIGHT,
    "hash_algorithm": "SHA-256"
}

stored_digest = create_payload_digest(encrypted_video)

nft_metadata["encrypted_payload_sha256"] = stored_digest

print("NFT metadata:")
print(json.dumps(nft_metadata, indent=2))

print(
    "Original encrypted payload integrity:",
    verify_payload_digest(encrypted_video, stored_digest)
)

# Tampering simulation: change one encrypted byte.
tampered_video = encrypted_video.copy()
tampered_video[0, 0, 0, 0] ^= np.uint8(1)

print(
    "Tampered payload integrity:",
    verify_payload_digest(tampered_video, stored_digest)
)

assert verify_payload_digest(encrypted_video, stored_digest)
assert not verify_payload_digest(tampered_video, stored_digest)
```

```python
class NFTAccessDemo:
    def __init__(self, token_id, initial_owner):
        self.token_id = token_id
        self.owner = initial_owner

    def transfer(self, current_owner, new_owner):
        if current_owner != self.owner:
            raise PermissionError("Only the current owner can transfer.")

        self.owner = new_owner

    def authorize(self, wallet_address):
        return wallet_address == self.owner


nft = NFTAccessDemo(
    token_id=1001,
    initial_owner="0xOWNER_A"
)

print("Owner A authorized:", nft.authorize("0xOWNER_A"))
print("Owner B authorized:", nft.authorize("0xOWNER_B"))

nft.transfer("0xOWNER_A", "0xOWNER_B")

print("After transfer:")
print("Owner A authorized:", nft.authorize("0xOWNER_A"))
print("Owner B authorized:", nft.authorize("0xOWNER_B"))

assert not nft.authorize("0xOWNER_A")
assert nft.authorize("0xOWNER_B")
```
